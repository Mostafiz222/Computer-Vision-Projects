```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt


model = YOLO("models/best.pt")


image_path = "test_images/road.jpg"


results = model.predict(
    source=image_path,
    conf=0.4
)


output = results[0].plot()


plt.figure(figsize=(10,8))

plt.imshow(
    cv2.cvtColor(
        output,
        cv2.COLOR_BGR2RGB
    )
)

plt.axis("off")
plt.show()
```
